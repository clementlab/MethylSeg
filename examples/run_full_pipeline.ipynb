# Run the Full MethylSeg Pipeline

This notebook trains new WGBS and TCGA/HM450K models, calls regions, cleans them, and writes summary outputs.

## Input used in this example

MethylSeg accepts tab-delimited `.tsv` or `.tsv.gz` files. WGBS count input
uses `CpG_chrm`, `CpG_beg`, `CpG_end`, `meth`, and `coverage`:

```text
CpG_chrm	CpG_beg	CpG_end	meth	coverage
chr1	10468	10469	7	12
chr1	10470	10471	3	10
```

TCGA/HM450K input uses beta values directly; `probe` is optional:

```text
CpG_chrm	CpG_beg	CpG_end	beta	probe
chr1	10468	10469	0.583	cg00000029
chr1	10470	10471	0.271	cg00000108
```

The complete inputs below are in `data/reference_files/` after running
`methylseg download_data_files`.


## Parameters

````python
from pathlib import Path

from methylseg import MethylSegPathway, MethylStateAssignmentMethod, MethylationStates
from methylseg.helper_classes import DATA_DIR

REFERENCE_DIR = DATA_DIR / "reference_files"
OUTPUT_ROOT = Path("methylseg_examples_output")
CHROM = "chr1"

WGBS_SAMPLE = REFERENCE_DIR / "WGBS_colon-primary-tumor_1_wgbs.tsv.gz"
TCGA_SAMPLE = REFERENCE_DIR / "TCGA-BD-A3EP-01A_450k.tsv.gz"
CHROMS = [CHROM]  # Use None for all chromosomes.
````

## WGBS full run

````python
wgbs_info, _ = MethylSegPathway.prepare_sample_info(
    sample_name="WGBS_colon-primary-tumor_1", sample_file=WGBS_SAMPLE,
    resolution="wgbs", min_coverage=10,
)
wgbs_pathway = MethylSegPathway(
    train_sample_info=wgbs_info, out_dir=OUTPUT_ROOT / "full_pipeline_wgbs"
)
wgbs_summary_paths = wgbs_pathway.run_pathway(
    sample_info=wgbs_info, chroms=CHROMS
)
wgbs_summary_paths
````

## TCGA HM450K full run

````python
tcga_info, _ = MethylSegPathway.prepare_sample_info(
    sample_name="TCGA-BD-A3EP-01A", sample_file=TCGA_SAMPLE, resolution="450k"
)
tcga_pathway = MethylSegPathway(
    train_sample_info=tcga_info, out_dir=OUTPUT_ROOT / "full_pipeline_tcga"
)
tcga_summary_paths = tcga_pathway.run_pathway(
    sample_info=tcga_info, chroms=CHROMS
)
tcga_summary_paths
````

## Outputs

Each returned list contains written raw and cleaned summary BED paths. Per-chromosome calls and `clean_regions/` metadata are stored under the corresponding output directory.
